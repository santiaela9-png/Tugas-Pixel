```python
import os
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Nama file gambar
lokasi_gambar = os.path.join(
    "/content/sample_data/gambar1.jpeg",
)

# Membaca gambar
if not os.path.exists(lokasi_gambar):
    print("Gambar tidak ditemukan:", lokasi_gambar)
    print("Pastikan gambar berada di folder gambar.")
    raise SystemExit

gambar = Image.open(lokasi_gambar).convert("RGB")
pixel = np.array(gambar)

print("=== INFORMASI GAMBAR ===")
print("Ukuran gambar :", gambar.size)
print("Total pixel   :", gambar.width * gambar.height)
print("Warna RGB     :", gambar.getpixel((100, 100)))

# Menampilkan gambar asli
plt.figure(figsize=(7, 8))
plt.imshow(gambar)
plt.title("Gambar Asli")
plt.axis("off")
plt.show()
```

```python
# MENAMPILKAN GAMBAR DALAM BENTUK PIXEL
plt.figure(figsize=(7, 6))
plt.imshow(pixel)
plt.title("Visualisasi Susunan Pixel")
plt.axis("off")
plt.show()

# MENAMPILKAN PEMBESARAN PIXEL
plt.figure(figsize=(7, 6))
plt.imshow(pixel[:50, :50])
plt.title("Pembesaran Pixel (50 x 50)")
plt.xticks(range(0, 50, 5))
plt.yticks(range(0, 50, 5))
plt.grid(color="white", linewidth=0.5)
plt.show()
```

```python
# HISTOGRAM WARNA RGB
plt.figure(figsize=(8, 5))
warna = ["red", "green", "blue"]

for i, nama in enumerate(warna):
    plt.hist(pixel[:, :, i].ravel(), bins=256,
             color=nama, alpha=0.4, label=nama.upper())

plt.title("Distribusi Intensitas Warna RGB")
plt.xlabel("Intensitas Warna (0–255)")
plt.ylabel("Jumlah Pixel")
plt.legend()
plt.show()
```

```python

# MENAMPILKAN WARNA DOMINAN
from collections import Counter

pixel_kecil = gambar.resize((100, 100))
warna_pixel = list(pixel_kecil.getdata())
warna_dominan = Counter(warna_pixel).most_common(5)

print("=== 5 WARNA DOMINAN ===")
for warna_rgb, jumlah in warna_dominan:
    print(f"RGB: {warna_rgb} | Jumlah pixel: {jumlah}")

# Visualisasi warna dominan
plt.figure(figsize=(8, 2))
for i, (warna_rgb, jumlah) in enumerate(warna_dominan):
    plt.bar(i, 1, color=np.array(warna_rgb) / 255)
    plt.text(i, 0.5, str(warna_rgb), ha="center", va="center",
             fontsize=9)

plt.title("Visualisasi 5 Warna Dominan")
plt.axis("off")
plt.show()
```
